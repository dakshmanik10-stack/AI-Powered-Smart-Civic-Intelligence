```python
import pandas as pd
import matplotlib.pyplot as plt
import re

# 1. Data Load & EDA
df = pd.read_csv('dataset/municipal_training_set_1100.csv')
print("Dataset Size:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())

# 2. Simple Visualization
plt.figure(figsize=(10,4))
df['municipal_department'].value_counts().head(5).plot(kind='bar', color='teal')
plt.title('Top 5 Departments by Complaints')
plt.show()

# 3. Text Preprocessing (Cleaning)
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text

df['cleaned_text'] = df['issue_description'].apply(clean_text)
print("\nOriginal vs Cleaned Data:")
print(df[['issue_description', 'cleaned_text']].head())
```
